# **Lab 1 - Introduction to Python**

Python is a *dynamic* and *interpreted* programming language. This means that variables of any type (numbers, characters, lists, etc.) can be created without the need to specify their type, and the type of an object can *change* during various operations.

Like in other programming languages, a code written in Python is composed of a sequence of instructions. By convention, **each line of code is an instruction**.
</br>
</br>
There are two possible working environments to write (and then execute) Python instructions:

1. Python **scripts** (files ".py"). They are the standard working environment. By launching a script, all the instructions inside it are executed (in sequence). We will see examples of this later.

2. Python **notebooks** (files ".ipynb"). Useful for educational and illustrative purposes, notebooks allow the initiation of a working session during which the user can execute various instructions, usually grouped into *cells*. The order of execution is determined by the user.
In particular, unlike scripts, notebooks allow the user to *interact* with the software. The session remains active until the user interrupts or restarts the runtime.

What we are using now is a notebook. There are *text cells* (like this one) and *code cells*, like the one below. Code cells can be executed with a mouse click or by pressing "shift+enter".

Within the code, using the "#" symbol, we can also insert **comment lines**. They do not constitute instructions to be executed but are helpful for those reading the code.

# **Control Flow**

## *For* loop
The for loop is a way to ensure that certain instructions are repeated for a predetermined number of times. Each iteration is driven by a *counter*, which varies within a predefined range of values.

As with functions, the entire body of the for loop must be written **below** the loop declaration and **indented**.

```python
# Defining a FOR loop




```

In general, the counter of a for loop can vary within any *iterable* object: a list, a tuple, etc. In particular, there is a way to construct a standard iterator using the **range** command.

```python
z = 1

for i in range(5): # <-- iterates the instructions 5 times.
                   #     ATTENTION: the counter "i" will start counting from "0" (thus stopping at 4).
  z = 2 * z
  print("Counter i = %d, value of z = %d." % (i, z))

print("\nWe are done. Now z = %d." % z)
```

## Conditional statements: *if* and *else* commands
Other times, we want the code to execute certain instructions only under certain circumstances. In particular, if a condition is met (if), we want it to do one thing; otherwise (else), we want it to do another.

Again, it is important to **pay attention to the indentation of the code**, as it determines which instructions fall under a branch and which do not.

```python
# Defining a conditional block of type IF - ELSE




```

To handle multiple conditions simultaneously, you can use the **and** and **or** keywords.

```python
x, y = 4, 9

if(x > 0 and y > 0):
  print("Both x and y are positive.")
else:
  print("At least one between x and y is negative.")
```

```python
if(x > 0 or y > 0):
  print("At least one between x and y is positive.")
else:
  print("Both x and y are negative.")
```

## Drawing graphs with **matplotlib**

A well-known library for graphs visualization is **matplotlib**. In particular, its submodule **matplotlib.pyplot**.

Given that the name is lenghty, people usually import it with a suitable pseudonym (typically **plt**).</br>
Note: numpy also has a classical pseudonym, which is **np**.

```python
import matplotlib.pyplot as plt
```

The main functions we are interested in are
- **figure**: creates an empty figure (not mandatory, but it allows choosing stuff such as the figure size);
- **plot**: draw lines / points with certain coordinates;
- **legend**: add a legend;
- **title**: add a title;
- **show**: show the graph (not mandatory, but useful for suppressing unwanted outputs).

Obviously there are many others, but we will see them during the course.

```python
# Example: drawing the graph of a given function (isolated points, piece-wise linear plot, etc.)




```

```python
# We also draw multiple graphs by calling "plot" repeatedly.
# In this case, it is good practice adding a label and a legend.

plt.figure(figsize = (5, 3.5))
plt.plot(x, y, '.', label = 'Points')
plt.plot(x, y, label = 'Continuous curve')
plt.title("Graph of $y=\sin(x)$")
plt.legend() # <-- activates the legend
plt.show()
```
