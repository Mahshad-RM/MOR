# **Lab 1 - Introduction to Python**

Python is a *dynamic* and *interpreted* programming language. This means that variables of any type (numbers, characters, lists, etc.) can be created without the need to specify their type, and the type of an object can *change* during various operations.

Like in other programming languages, a code written in Python is composed of a sequence of instructions. By convention, **each line of code is an instruction**.
</br>
</br>
There are two possible working environments to write (and then execute) Python instructions:

1. Python **scripts** (files ".py"). They are the standard working environment. By launching a script, all the instructions inside it are executed (in sequence). We will see examples of this later.

2. Python **notebooks** (files ".ipynb"). Useful for educational and illustrative purposes, notebooks allow the initiation of a working session during which the user can execute various instructions, usually grouped into *cells*. The order of execution is determined by the user.
In particular, unlike scripts, notebooks allow the user to *interact* with the software. The session remains active until the user interrupts or restarts the runtime.

What we are using now is a notebook. There are *text cells* (like this one) and *code cells*, like the one below. Code cells can be executed with a mouse click or by pressing "shift+enter".

## Functions

Another very useful concept is that of **function**. In Python, a function is an object that produces outputs based on certain inputs.
</br></br>
To create a function, the **def** command is used, followed by the list of required inputs (in parentheses) and the sequence of operations that define it.

```python
def triangleArea(basis, height):   # <-- This function is called 'triangleArea' and it accepts two inputs: basis and height
  return 0.5*basis*height
```

```python
triangleArea(4, 5)
```

Note: all instructions that are part of the *body* of the function must be written **below** the definition of the function itself and followed by appropriate **indentation**.

A function can perform various calculations within its body: the only results that will be returned (and visible) are those preceded by the **return** command.

```python
def triRectPerimeter(cathetus1, cathetus2):
  hypotenuse = (cathetus1**2 + cathetus2**2)**0.5    # this line is part of the function's body
  perimeter = cathetus1 + cathetus2 + hypotenuse     # this line is part of the function's body
  return perimeter                                   # this line is part of the function's body

# The following lines are NOT part of the function's body
x = 3.0
y = 4.0
triRectPerimeter(x, y)
```

**Note**: By leveraging the concept of a tuple, you can build functions that return multiple outputs simultaneously.

```python
def sumAndproduct(a, b):
  return a+b, a*b

x, y = sumAndproduct(3, 5)
print(x)
print(y)
```

Finally, when defining a function, we can also decide that some inputs are *optional*: the user can pass a value, but it is not mandatory. If not provided, a default value will be used. The syntax is as follows:

```python
def summation(a, b, c = 0):
  return a + b + c
```

```python
summation(4, 2, 8)
```

```python
summation(4, 2)
```

# **Control Flow**

## *For* loop
The for loop is a way to ensure that certain instructions are repeated for a predetermined number of times. Each iteration is driven by a *counter*, which varies within a predefined range of values.

As with functions, the entire body of the for loop must be written **below** the loop declaration and **indented**.

```python
z = 1

# The code below will be repeated 3 times, during which the 'i' counter will take on
# the values 1, 2, and 3.

for i in [1, 2, 3]:
  z = 2 * z
  print("Now, we are in loop i = %d, and z equals %d." % (i, z)) # <-- using tuples, we can format multiple values!

print("\nWe are done. Now z = %d." % z) # <-- this instruction is NOT part of the FOR loop, so it will not be repeated.
```

In general, the counter of a for loop can vary within any *iterable* object: a list, a tuple, etc. In particular, there is a way to construct a standard iterator using the **range** command.

```python
z = 1

for i in range(5): # <-- iterates the instructions 5 times.
                   #     ATTENTION: the counter "i" will start counting from "0" (thus stopping at 4).
  z = 2 * z
  print("Counter i = %d, value of z = %d." % (i, z))

print("\nWe are done. Now z = %d." % z)
```

## *While* loop
Sometimes, it is useful to **repeat a set of instructions as long as a certain condition is met**. In particular, we want to repeat certain operations, but we do not know in advance how many times we want to do it (otherwise, we would use a for loop!).

To achieve this, there is the *while* loop. In its declaration, the condition that determines the execution of the loop must be specified.

```python
x = 5
while (x > 0): # <-- as long as x is greater than zero, the instructions inside the WHILE loop will be repeated
  x = x - 2
  print("x is still positive: subtracting 2 and checking again.")

print("\nDone!") # <-- this instruction is NOT part of the WHILE loop
```

## Conditional statements: *if* and *else* commands
Other times, we want the code to execute certain instructions only under certain circumstances. In particular, if a condition is met (if), we want it to do one thing; otherwise (else), we want it to do another.

Again, it is important to **pay attention to the indentation of the code**, as it determines which instructions fall under a branch and which do not.

```python
x = 5
y = 3

if(x > y):
  print("x is greater than y.")
else:
  print("x is less than or equal to y.")
```

To handle multiple conditions simultaneously, you can use the **and** and **or** keywords.

```python
x, y = 4, 9

if(x > 0 and y > 0):
  print("Both x and y are positive.")
else:
  print("At least one between x and y is negative.")
```

```python
if(x > 0 or y > 0):
  print("At least one between x and y is positive.")
else:
  print("Both x and y are negative.")
```

<mark>**Exercise 6**</mark></br>
Create a function that, given a matrix $A$ and a nonnegative integer $n\ge0$, compute the $n$th power of $A$ *in the matrix sense*.

```python
# Exercise 6

def MatPow(A, n):
  d = A.shape[0] # <-- size of A (assuming it's a square matrix)
  B = numpy.eye(d)
  for i in range(n):
    B = B @ A
  return B


Z = numpy.array([[0, 1],
                [-1, 0]])
MatPow(Z, 2)
```

## Drawing graphs with **matplotlib**

A well-known library for graphs visualization is **matplotlib**. In particular, its submodule **matplotlib.pyplot**.

Given that the name is lenghty, people usually import it with a suitable pseudonym (typically **plt**).</br>
Note: numpy also has a classical pseudonym, which is **np**.

```python
import matplotlib.pyplot as plt
```

The main functions we are interested in are
- **figure**: creates an empty figure (not mandatory, but it allows choosing stuff such as the figure size);
- **plot**: draw lines / points with certain coordinates;
- **legend**: add a legend;
- **title**: add a title;
- **show**: show the graph (not mandatory, but useful for suppressing unwanted outputs).

Obviously there are many others, but we will see them during the course.

```python
x = numpy.linspace(0, 2*numpy.pi, 100)
y = numpy.sin(x)

plt.figure(figsize = (5, 3.5)) # <-- create a plot of width 5 and height 3.5
plt.plot(x, y, '.')          # <-- draws dots at the given coordinates
plt.title("Graph of $y=\sin(x)$") # <-- title (TeX is supported!)
plt.show()
```

```python
# If, while calling "plot", we do not specify any marker type, matplotlib will join points
# through a series of lines (piece-wise linear interpolation).
# If enough points are provided, this gives the illusion of a smooth curve.

plt.figure(figsize = (5, 3.5))
plt.plot(x, y)
plt.title("Graph of $y=\sin(x)$")
plt.show()
```

```python
# We also draw multiple graphs by calling "plot" repeatedly.
# In this case, it is good practice adding a label and a legend.

plt.figure(figsize = (5, 3.5))
plt.plot(x, y, '.', label = 'Points')
plt.plot(x, y, label = 'Continuous curve')
plt.title("Graph of $y=\sin(x)$")
plt.legend() # <-- activates the legend
plt.show()
```
